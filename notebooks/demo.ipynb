```python
import numpy as np

from spectral_clustering.graphs import (
    pairwise_distance,
    rbf_affinity,
    degree_matrix,
    unnormalized_laplacian,
)

X = np.array([
    [0, 0],
    [3, 0],
    [0, 4],
])

distances = pairwise_distance(X)
W = rbf_affinity(X, gamma=0.1)
D = degree_matrix(W)
L = unnormalized_laplacian(W)

print("Distances =\n", distances)
print("W =\n", W)
print("D =\n", D)
print("L =\n", L)

# Đáp án khoảng cách tính tay: tam giác 3–4–5
assert np.allclose(distances, [
    [0, 3, 4],
    [3, 0, 5],
    [4, 5, 0],
])

# Kiểm tra công thức RBF và các tính chất ma trận
assert np.isclose(W[0, 1], np.exp(-0.9))
assert np.allclose(W, W.T)
assert np.all(W >= 0)
assert np.allclose(np.diag(W), 0)
assert np.allclose(np.diag(D), W.sum(axis=1))
assert np.allclose(L, D - W)
assert np.allclose(L.sum(axis=1), 0)

print("Checks passed")
```

```python
import numpy as np
from sklearn.datasets import make_moons
from sklearn.preprocessing import StandardScaler
from spectral_clustering.spectral import spectral_clustering

# Tạo 300 điểm thuộc hai hình trăng lưỡi liềm
X, _ = make_moons(n_samples=300, noise=0.05, random_state=42)
X_scaled = StandardScaler().fit_transform(X)

# Chạy hàm Spectral-RBF của nhóm
labels = spectral_clustering(X_scaled, n_clusters=2, gamma=5.0, seed=42)

cluster_ids, counts = np.unique(labels, return_counts=True)
print("Kích thước X:", X_scaled.shape)
print("Kích thước labels:", labels.shape)
print("Số điểm từng cụm:", dict(zip(cluster_ids.tolist(), counts.tolist())))

assert labels.shape == (300,)
assert len(cluster_ids) == 2
print("Moons: chạy thành công trên 300 điểm, thu được 2 cụm.")
```
